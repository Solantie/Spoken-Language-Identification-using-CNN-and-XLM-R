Absolute Paths:

- Cantonese: /resources/speech/corpora/fleurs/yue_hant_hk
- Japanese: /resources/speech/corpora/fleurs/ja_jp
- German: /resources/speech/corpora/fleurs/de_de
- Dutch: /resources/speech/corpora/fleurs/nl_nl
- American English: /resources/speech/corpora/fleurs/en_us

## Metadata structure
id, audio_id, audio_path, lang_id, language, lang_group_id, lang_group, gender, split, num_samples, sampling_rate, duration

- id: automatically assigned
- audio_id: extracted from raw tsv file
- audio_path: absolute path on IMS server
- lang_id: mapped from language
    - Mandarin: 0
    - Cantonese: 1
    - Japanese: 2
    - German: 3
    - Dutch: 4
    - American English: 5
- language: extracted from raw folder name
- lang_group_id:
    - CJK: 0
    - WE: 1
- lang_group: mapped from language
- gender: extracted from raw tsv file
- split: extracted from raw tsv file name
- num_samples: extracted from raw tsv file
- sampling_rate: 16000 according to (https://arxiv.org/pdf/2205.12446)
- duration: `num_samples / sampling_rate`

An example of a data instance of the config `en_us`:

```json
{
  "id": 151,
  "audio_id": "11559549184357409250.wav",
  "audio_path": "/resources/speech/corpora/fleurs/en_us",
  "lang_id": 5,
  "language": "en_us",
  "lang_group_id": 1,
  "lang_group": "WE",
  "gender": "female",
  "split": "train",
  "num_samples": 120000,
  "sampling_rate": 16000,
  "duration": 7.5
}
```
